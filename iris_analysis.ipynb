```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

%matplotlib inline
```

```python
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
df.head()
```

```python
df.groupby('species').mean(numeric_only=True)
```

```python
plt.figure(figsize=(10, 6), dpi=150)

df.boxplot(column='petal length (cm)', by='species', grid=False)
plt.title('Petal Length by Species')
plt.suptitle('')  # 去掉默认标题
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.tight_layout()
plt.savefig('petal_length_boxplot.png', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10, 6), dpi=150)

colors = {'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'}
for species, color in colors.items():
    subset = df[df['species'] == species]
    plt.scatter(subset['petal length (cm)'], subset['petal width (cm)'],
                label=species, color=color, alpha=0.7)

plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Petal Length vs Petal Width by Species')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('petal_scatter.png', bbox_inches='tight')
plt.show()
```
